```python
# Step 1: Installing PySpark
# PySpark is the Python API for Apache Spark.
# It allows us to perform distributed data processing using Python.

pip install pyspark
```

```python
# Step 2: Creating SparkContext
# SparkContext is the entry point for working with RDDs in Apache Spark.
# getOrCreate() checks whether an existing SparkContext is available.
# If it exists, it reuses it; otherwise, it creates a new one.

import pyspark
sc=pyspark.SparkContext.getOrCreate()
print("Spark is ready")
```

```python
# Step 3: Uploading the Input File
# The files module from Google Colab is used to upload files from the local system.
# files.upload() allows us to select and upload the text file required for processing.
from google.colab import files
data = files.upload()
```

```python
# Step 4: Creating RDD from the Input File
# textFile() reads the uploaded text file and creates an RDD.
# Each line of the file is treated as an individual RDD element.
# take(5) retrieves the first five lines from the RDD.
# count() returns the total number of lines in the file.
# Retrieving data is an action, and count() is also an action.

lines = sc.textFile("/content/book-war-and-peace.txt")
print(lines.take(5))
print(lines.count())
```

```python
# Step 5: Removing Empty Lines
# filter() is a transformation used to select elements based on a condition.
# The lambda function checks the length of each line.
# Only lines with a length greater than 0 are retained.
# Empty lines are removed from the RDD.


noNullLines = lines.filter(lambda line: len(line) > 0)
```

```python
# Step 6: Splitting Lines into Words
# flatMap() is a transformation that converts each input element into multiple output elements.
# The split() function splits each line into individual words based on whitespace.
# Each word becomes a separate element in the RDD.


words = noNullLines.flatMap(lambda line: line.split(" "))
```

```python
# Step 7: Converting Words into Uppercase
# map() is a transformation that applies a function to each element of the RDD.
# The upper() function converts each word into uppercase letters.
# The transformed words are stored in a new RDD.

upperWords = words.map(lambda word:word.upper())
```

```python
# Step 8: Creating Word-Count Pairs
# map() is used to convert each word into a key-value pair.
# Each word is assigned a count of 1.
# The output is stored as (word, 1) pairs in the RDD.

pairedOnes = upperWords.map(lambda uw: (uw, 1))
```

```python
# Step 9: Counting Word Frequencies
# reduceByKey() is a transformation that groups values with the same key.
# It adds the counts of each word using the lambda function.
# It performs local aggregation and shuffling to combine values efficiently.
# The final word counts are stored in the wordCounts RDD

wordCounts=pairedOnes.reduceByKey(lambda prev,next:prev+next)
```

```python
# Step 10 : Final Word Count
# take(5) retrieves the first five word-count pairs.
# print() displays the retrieved results.

for word in wordCounts.take(5):
  print("",word)
```
